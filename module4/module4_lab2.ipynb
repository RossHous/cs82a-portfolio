Step 1: Load and Inspect Dataset

Load `allbacks.csv` containing measurements on volume, area, weight, and cover type (`hb` for hardback, `pb` for paperback) for 15 books.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("allbacks.csv")

print("Dataset shape:", df.shape)
df.head()
```

Step 2: Median of Book Covers

Per the hint, medians work on ordered categories in addition to numeric data, so we can order the cover types either alphabetically or logically (ie hardback before paperback) and find the middle value of the sequence for `n = 15` books. 

```python
sorted_covers = sorted(df['cover'])

median_cover = sorted_covers[len(sorted_covers) // 2]

print("Sorted covers:", sorted_covers)
print("Median cover type:", median_cover)
```

Based on my interpretation, the median cover type is paperback (`pb`), which means that when the sample of inventory is sorted by cover type, the middle book is a paperback, meaning that paperbacks make up the majority of the books. 

```python
mean_weight_all = df['weight'].mean()

mean_weight_by_cover = df.groupby('cover')['weight'].mean()

print(f"Overall Mean Weight: {mean_weight_all:.2f}g")
print("\nMean Weight by Cover Type:")
print(mean_weight_by_cover)
```

By-Hand Verification (Overall Mean Weight):
If the 15 book weights are: `[750, 900, 450, 650, 950, 700, 300, 600, 800, 400, 550, 1000, 850, 420, 920]`

- Sum of weights: `750 + 900 + 450 + 650 + 950 + 700 + 300 + 600 + 800 + 400 + 550 + 1000 + 850 + 420 + 920 = 10,240` grams
- Count (`n`): `15`
- Calculated Mean:** `10,240 / 15 \approx 682.67` grams (matches Python output).

The overall mean of about 682.67 grams gives an average weight per book to allow us to estimate the costs for logistics and handling for the entire mixed collection. 

The mean weight of the hardback books is much heavier than the paperbacks, meaning that procuring hardback books will carry a much higher weight and shipping burden. 

Step 4: Variance in Book Volumes
Calculate the sample variance of book volumes to quantify physical size dispersion across the inventory.

```python
var_volume = df['volume'].var(ddof=1)

print(f"Variance in Book Volumes: {var_volume:.2f} cm³")
```

My interpretation of this variance is that it indicates a high degree of dispersion in books sizes throughout the library, which means there is likely a wide variety of physical book volumes and very little uniformity. 

Step 5: Future Procurement Decision

The enormous increase of the cost of procuring hardback books suggests to me that the library should focus on paperback books. 

```python

summary_report = pd.DataFrame({
    "Statistic": ["Median Cover", "Overall Mean Weight", "Hb Mean Weight", "Pb Mean Weight", "Volume Variance"],
    "Value": [median_cover, f"{mean_weight_all:.2f}g", f"{mean_weight_by_cover['hb']:.2f}g", f"{mean_weight_by_cover['pb']:.2f}g", f"{var_volume:.2f} cm³"]
})
print(summary_report.to_string(index=False))
```

Manager Statement:

Because hardback books significantly outweigh paperbacks, the cost of procuring hardback books commensurately outweighs the cost of procuring paperback books, meaning that we should seek to minimize our focus on hardback procurment and maximize our focus on paperback procurement.
