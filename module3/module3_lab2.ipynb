```python
import pandas as pd

df = pd.read_csv("usarrests.csv")

print("--- Data Preview ---")
display(df.head())

print("\n--- Summary Statistics ---")
display(df.describe())
```

```python
print("--- Checking for Impossible Urban Population (> 100) ---")
impossible_urban = df[df["UrbanPop"] > 100]
display(impossible_urban)

print("\n--- Checking for Extreme Crime Rates ---")
extreme_crime = df[(df["Assault"] > 300) | (df["Murder"] > 20)]
display(extreme_crime)
```

Outlier Investigation Log: US Arrests Data Analysis: Impossible vs. Extreme

For Impossible Values (UrbanPop > 100%):
Given that we're looking at a poopulation of people, the maximum value we can have is 100%, so any value over 100 is impossible as opposed to an outlier, and is more likely to be an error or evidence of corruption. As a result, any rows showing a value over 100 should be investigated and corrected if possible. They should be dropped if they cannot be corrected, as known mistakes should not influence other data. 

For Extreme Values (Extremely High Assault/Murder Rates):
In contrast, extremely high arrest rates (per 100,000 residents), while rare, are not impossible. Extreme values should not be automatically deleted, as that can destroy legitimate variance and ruin bias checks. As a result, these rows should be retained, but carefully documented to show that they represent outliers, not mistakes.
