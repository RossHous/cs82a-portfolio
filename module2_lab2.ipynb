```python
def multiply_for(a: int, b: int) -> int:
    result = 0
    is_negative = (a < 0) ^ (b < 0)
    for _ in range(abs(b)):
        result += abs(a)

    return -result if is_negative else result

test_cases = [(6, 7), (-4, 5), (3, -8), (-9, -2), (0, 10)]

print("Testing multiply_for:")
for a, b in test_cases:
    result = multiply_for(a, b)
    expected = a * b
    assert result == expected, f"Failed for {a} * {b}"
    print(f"  {a} * {b} = {result} (Expected: {expected})")
```

```python
def multiply_while(a: int, b: int) -> int:
    result = 0
    count = 0
    is_negative = (a < 0) ^ (b < 0)

    while count < abs(b):
        result += abs(a)
        count += 1

    return -result if is_negative else result

test_cases = [(1, 7), (-4, 5), (3, -8), (-9, -2), (0, 10)]

print("Testing multiply_while:")
for a, b in test_cases:
    result = multiply_while(a, b)
    expected = a * b
    assert result == expected, f"Failed for {a} * {b}"
    print(f"  {a} * {b} = {result} (Expected: {expected})")
```

A for loop has built-in termination because it covers a set number of variables. If we already know we have to check 10 variables, we can put them in a range and run the for loop to cover the entire range. However, if we don't know how many variables we have, we can use a while loop, which repeats an action for as long as a condition remains TRUE, and terminates once the condition becomes FALSE. If a termination condition is not included in a while loop, it will run forever!
