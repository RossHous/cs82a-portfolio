```python
def convert_lbs_to_kg(pounds):
    # convert a weight in pounds (lbs) to kilograms (kgs)
    conversion_factor = 0.453592
    kg = pounds * conversion_factor
    return kg

# first call: test value
weight_1 = 150
print(f"{weight_1} lbs is {convert_lbs_to_kg(weight_1):.2f} kg")

# second call: a different test value
weight_2 = 200
print(f"{weight_2} lbs is {convert_lbs_to_kg(weight_2):.2f} kg")

# third call: my own weight in lbs
my_weight_lbs = 260
print(f"My weight of {my_weight_lbs} lbs is {convert_lbs_to_kg(my_weight_lbs):.2f} kg")
```
