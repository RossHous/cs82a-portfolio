rules: 

If the temperature is LESS THAN 0 degrees, it is categorized as "Freezing".

If the temperature is BETWEEN 0 and 10 degrees, it is categorized as "Cold".

If the temperature is BETWEEN 11 and 20 degrees, it is categorized as "Mild"

If the temperature is BETWEEN 21 and 30 degrees, it is categorized as "Warm".

If the temperature is MORE THAN 30 degrees, it is categorized as "Hot".

```python
temperatures = [10, 16, 23, 28, 14, 22, 30]

for temp in temperatures:
    if temp < 0:
        category = "Freezing"
    elif 1 <= temp <= 10:
        category = "Cold"
    elif 11 <= temp <= 20:
        category = "Mild"
    elif 21 <= temp <= 30:
        category = "Warm"
    elif temp > 30:
        category = "Hot"

    print(f"{temp}°C: {category}")
```
