Create and insert (Appendix C.2)

Following C.2’s example, create a customers table with at least four columns (id, name, city, and one numeric column) using a SQL CREATE TABLE statement, then INSERT at least eight fictional rows. MySQL’s types map to SQLite almost directly: use INTEGER, TEXT, REAL.

Deliverable: the CREATE and INSERT statements running via con.execute(), and a SELECT * proving the rows exist.

```python
import sqlite3
import pandas as pd

con = sqlite3.connect('lab2.db')
cursor = con.cursor()

cursor.execute("DROP TABLE IF EXISTS customers;")

cursor.execute("""
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    loyalty_score REAL
);
""")

fictional_customers = [
    (1, 'Caleb Hearon', 'Los Angeles', 4.8),
    (2, 'Michael Showalter', 'Santa Monica', 3.5),
    (3, 'Tim Robinson', 'San Francisco', 4.2),
    (4, 'Sam Richardson', 'Seattle', 4.9),
    (5, 'Evan Wright', 'Austin', 2.7),
    (6, 'Seth Rogen', 'Chicago', 3.9),
    (7, 'George Miller', 'New York', 4.1),
    (8, 'Jennifer Connolley', 'Boston', 4.6)
]

cursor.executemany("""
INSERT INTO customers (id, name, city, loyalty_score) 
VALUES (?, ?, ?, ?);
""", fictional_customers)

con.commit()

df_lab2 = pd.read_sql("SELECT * FROM customers;", con)
df_lab2
```
