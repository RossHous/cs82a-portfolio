Hands-On Problem 2.2 Stated Assumption:
Looking at Google Maps for the region surrounding Pittsburgh, and doing some research on the major rivers in the area, it seems reasonable to assume that these bridges need to be a minimum of 50 feet long. On the other end, a quick Google search says that the widest river in all of Pennsylvania is about a mile wide, and that's not near Pittsburgh, so it seems reasonable to assume that no bridges in the Pittsburgh area will be longer than about 5000 feet (a mile is 5,280 feet, but 5000 feet just seems like an easier boundary). 

Plausible Range: 50 feet \le length \le 5,000 feet

Plausible Range: $50 \text{ feet} \le \text{length} \le 5,000 \text{ feet}$

```python
import pandas as pd

df_bridges = pd.read_csv("bridges.csv")

print("--- Column Names ---")
print(df_bridges.columns)

print("\n--- Summary Statistics for Length ---")

length_col = "LENGTH" if "LENGTH" in df_bridges.columns else "length"
display(df_bridges[length_col].describe())


print("\n--- Flagged Outliers Based on Assumption ---")
flagged_bridges = df_bridges[(df_bridges[length_col] < 50) | (df_bridges[length_col] > 5000)]
display(flagged_bridges)
```

Deliverable Log & Justification

Flagged Rows Result: The filter isolates any bridges falling outside our 50 ft to 5,000 ft threshold. 
Justification & Action:
Because we set our lower boundary at 50 feet to exclude things like pedestrian bridges and culverts, we can check any result under 50 feet against historical bridge archives. On the other hand, any result over 5,000 feet is more likely to be an error. Either way, we can easily check all the outliers, whether too small or too large, and document them for either verification/correction or separate them from the rest to avoid distorting the verified data. 

In this case, zero bridges were flagged as too long, meaning our data set is likely in good shape!
